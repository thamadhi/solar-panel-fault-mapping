```python
from ultralytics import YOLO
yaml_path = "dataset.yaml"
```

```python
# Initialize YOLOv8 model
model = YOLO("yolov8n-seg.pt")

# Start training
model.train(
    data="dataset.yaml",
    epochs=20,
    batch=32,
    imgsz=640,
    device="cpu",
    augment=True,
    project="runs/train",
    name="solar_panel_aug",
    exist_ok=True
)
```

```python
metrics = model.val(data="dataset.yaml", split="test")
```
