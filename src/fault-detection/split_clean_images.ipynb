```python
import os
import shutil
from sklearn.model_selection import train_test_split
```

```python
source_dir = "PV_Images/clean"
```

```python
# Output directories
train_dir = "PV_Images/train/clean"
test_dir = "PV_Images/test/clean"
valid_dir = "PV_Images/valid/clean"
```

```python
# Creating output folders
for d in [train_dir, test_dir, valid_dir]:
  os.makedirs(d, exist_ok=True)
```

```python
images = {img for img in os.listdir(source_dir) if os.path.isfile(os.path.join(source_dir, img))}
```

```python
# Split into train and temp
train_imgs, temp_imgs = train_test_split(images, test_size=0.3, random_state=42)
```

```python
test_imgs, valid_imgs = train_test_split(temp_imgs, test_size=0.5, random_state=42)
```

```python
def copy_files(file_list, dest_dir):
  """
  Used to copy images
  """
  for img in file_list:
    shutil.copy(os.path.join(source_dir, img), os.path.join(dest_dir, img))
```

```python
# Copy to respective folders
copy_files(train_imgs, train_dir)
copy_files(test_imgs, test_dir)
copy_files(valid_imgs, valid_dir)
```

```python
print("Train images:", len(train_imgs))
print("Test Images:", len(test_imgs))
print("Valid images:", len(valid_imgs))
```
